```python
# Load the Salesforce report exports from Excel files
ep_df = pd.read_excel('NewCompanySalesExistingBusinessUnits002.xlsx')  # Read default sheet for opportunities
```

```python
# read column names
print(ep_df.columns)
```

```python
# Count the number of unique customers for each Business Unit and CustomerGroup
grouped = ep_df.groupby(['BusinessUnitBasedOnWarehouse', 'CustomerGroup']).size().reset_index(name='count')

# Plot a stacked horizontal bar chart
fig = px.bar(grouped, 
             x='count', 
             y='BusinessUnitBasedOnWarehouse', 
             color='CustomerGroup', 
             orientation='h', 
             title='Total Customers per Business Unit (Stacked by CustomerGroup)', 
             labels={'count': 'Number of Customers', 'BusinessUnitBasedOnWarehouse': 'Business Unit'},
             width=1450, 
             height=1200)

# Update layout to stack bars
fig.update_layout(barmode='stack')
fig.show()
```

```python
# Group by BusinessUnitBasedOnWarehouse and CustomerGroup and then count unique CustomerNum values
grouped = ep_df.groupby('BusinessUnitBasedOnWarehouse')['CustomerNum'].nunique().reset_index(name='count')

# Plot the horizontal bar chart
fig = px.bar(grouped, 
             x='count', 
             y='BusinessUnitBasedOnWarehouse', 
             orientation='h', 
             title='Number of Unique Customers per Business Unit (based on unique CustomerNum)', 
             labels={'count': 'Number of Unique Customers', 'BusinessUnitBasedOnWarehouse': 'Business Unit'},
             width=1450, 
             height=400)

# Add total number of unique customers to the end of each bar
for i in range(len(grouped)):
    fig.add_annotation(x=grouped['count'][i], 
                       y=grouped['BusinessUnitBasedOnWarehouse'][i], 
                       text=str(grouped['count'][i]),  # Convert to string for annotation
                       showarrow=False, 
                       xanchor='left',  # Anchor the text at the end of the bar
                       yanchor='middle')

# Display the figure
fig.show()

# Export chart as an HTML file
fig.write_html("NumberofUniqueCustomersperBusinessUnit.html")


```

```python
# Count the number of unique CustomerNum values for each BusinessUnitBasedOnWarehouse
grouped = ep_df.groupby('BusinessUnitBasedOnWarehouse')['CustomerNum'].nunique().reset_index(name='count')

# Plot the horizontal bar chart
fig = px.bar(grouped, 
             x='count', 
             y='BusinessUnitBasedOnWarehouse', 
             orientation='h', 
             title='Number of Unique Customers per Business Unit (based on CustomerNum)', 
             labels={'count': 'Number of Customers', 'BusinessUnitBasedOnWarehouse': 'Business Unit'},
             width=1450, 
             height=400)

# Add total number of unique customers to the end of each bar
for i in range(len(grouped)):
    fig.add_annotation(x=grouped['count'][i], 
                       y=grouped['BusinessUnitBasedOnWarehouse'][i], 
                       text=str(grouped['count'][i]),  # Convert to string for annotation
                       showarrow=False, 
                       xanchor='left',  # Anchor the text at the end of the bar
                       yanchor='middle')

# Display the figure
fig.show()

```

```python
grouped_business_unit = ep_df.groupby('BusinessUnitBasedOnWarehouse')['YTD'].sum().reset_index(name='sum').sort_values(by='sum', ascending=False)

# Plotting a bar chart
fig = px.bar(grouped_business_unit, 
             x='sum', 
             y='BusinessUnitBasedOnWarehouse', 
             orientation='h', 
             title='YTD Revenue per Business Unit', 
             labels={'sum': 'YTD Revenue', 'BusinessUnitBasedOnWarehouse': 'Business Unit'},
             width=1450, 
             height=400)

fig.show()

# Export chart as an HTML file
fig.write_html("YTDRevenueperBusinessUnit.html")
```

```python
# Group the data by BusinessUnitBasedOnWarehouse and TAM/IAM/ADS, and sum the YTD values
grouped_sales_rep_unit = ep_df.groupby(['BusinessUnitBasedOnWarehouse', 'TAM/IAM/ADS'])['YTD'].sum().reset_index()

# Round the 'YTD' values to remove decimals and convert to integers
grouped_sales_rep_unit['YTD'] = grouped_sales_rep_unit['YTD'].round(0).astype(int)

# Plotting a bar chart with hover information
fig = px.bar(grouped_sales_rep_unit, 
             x='YTD', 
             y='BusinessUnitBasedOnWarehouse', 
             color='TAM/IAM/ADS', 
             title='Sales by Business Unit and TAM/IAM/ADS', 
             labels={'YTD': 'YTD Revenue', 'BusinessUnitBasedOnWarehouse': 'Business Unit'},
             width=1450, 
             height=800,
             hover_data={'YTD': ':.0f'},  # Show whole number YTD values on hover
             template='plotly')  # Optional: Can add 'plotly' template for better aesthetics

# Update the x-axis to show dots as thousand separators
fig.update_xaxes(tickformat=",.0f", tickprefix="$")  # Format with thousand separators and dollar sign

# Display the figure
fig.show()

# Export chart as an HTML file
fig.write_html("SalesbyBusinessUnitandSalesRep.html")
```

```python
# Count the number of occurrences of each CustomerGroup for each TAM/IAM/ADS (Sales Rep) and sort in descending order
grouped = ep_df.groupby('TAM/IAM/ADS')['CustomerGroup'].count().reset_index(name='count').sort_values(by='count', ascending=True)

# Plot the horizontal bar chart with size adjustments
fig = px.bar(grouped, 
             x='count', 
             y='TAM/IAM/ADS', 
             orientation='h', 
             title='Number of Customers per Sales Rep (TAM/IAM/ADS)', 
             labels={'count': 'Number of Customers', 'TAM/IAM/ADS': 'Sales Rep'},
             width=1450,  # Adjust the width
             height=900)  # Adjust the height

# Add total number of customers to the end of each bar
for i in range(len(grouped)):
    fig.add_annotation(x=grouped['count'][i], 
                       y=grouped['TAM/IAM/ADS'][i], 
                       text=str(grouped['count'][i]),  # Convert to string for annotation
                       showarrow=False, 
                       xanchor='left',  # Anchor the text at the end of the bar
                       yanchor='middle')

# Display the figure
fig.show()

# Export chart as an HTML file
fig.write_html("NumberofCustomersperSalesRep.html")
```

```python
# Replace missing values in the 'NAM' column with 'No NAM Assigned'
ep_df['NAM'] = ep_df['NAM'].fillna('No NAM Assigned')

# Count the number of occurrences of each CustomerGroup for each NAM
grouped = ep_df.groupby('NAM')['CustomerGroup'].count().reset_index(name='count').sort_values(by='count', ascending=True)

# Plot the horizontal bar chart with size adjustments
fig = px.bar(grouped, 
             x='count', 
             y='NAM', 
             orientation='h', 
             title='Number of Customers per NAM', 
             labels={'count': 'Number of Customers', 'NAM': 'NAM'},
             width=1450,  # Adjust the width
             height=400)  # Adjust the height

# Add total number of customers to the end of each bar
for i in range(len(grouped)):
    fig.add_annotation(x=grouped['count'][i], 
                       y=grouped['NAM'][i], 
                       text=str(grouped['count'][i]),  # Convert to string for annotation
                       showarrow=False, 
                       xanchor='left',  # Anchor the text at the end of the bar
                       yanchor='middle')

# Display the figure
fig.show()

# Export chart as an HTML file
fig.write_html("NumberofCustomersperNAM.html")
```

```python
grouped_customer = ep_df.groupby('CustomerName')['YTD'].sum().reset_index(name='sum').sort_values(by='sum', ascending=True)

# Plotting a bar chart
fig = px.bar(grouped_customer, 
             x='sum', 
             y='CustomerName', 
             orientation='h', 
             title='YTD Revenue per Customer Name', 
             labels={'sum': 'YTD Revenue', 'CustomerName': 'Customer'},
             width=1200, 
             height=2000)

fig.show()

```

```python
# Clean and convert YTD column to numeric
ep_df['YTD'] = ep_df['YTD'].replace(',', '', regex=True).astype(float)

# Group the data by CustomerGroup and sum the YTD values
grouped = ep_df.groupby('CustomerGroup')['YTD'].sum().reset_index(name='sum').sort_values(by='sum', ascending=True)

# Round the 'sum' values to remove decimals and convert to integers
grouped['sum'] = grouped['sum'].round(0).astype(int)

# Plot the horizontal bar chart
fig = px.bar(grouped, 
             x='sum', 
             y='CustomerGroup', 
             orientation='h', 
             title='YTD Revenue per Customer Group', 
             labels={'sum': 'YTD Revenue', 'CustomerGroup': 'Customer Group'},
             width=1450, 
             height=2400)

# Add annotations for YTD revenue at the end of each bar, with dots as thousand separators
for i in range(len(grouped)):
    fig.add_annotation(x=grouped['sum'][i], 
                       y=grouped['CustomerGroup'][i], 
                       text=f"${grouped['sum'][i]:,.0f}".replace(",", "."),  # Format with dots as thousand separators
                       showarrow=False, 
                       xanchor='left', 
                       yanchor='middle')

# Update the x-axis to show dots as thousand separators
fig.update_xaxes(tickformat=",.0f", tickprefix="$")  # Tick format with thousand separators and dollar sign

# Display the figure
fig.show()

# Export chart as an HTML file
fig.write_html("YTDRevenueperCustomerGroup.html")
```

```python
# Group the data by TAM/IAM/ADS and sum the YTD values, rounding to whole numbers
grouped_role = ep_df.groupby('TAM/IAM/ADS')['YTD'].sum().reset_index(name='sum').sort_values(by='sum', ascending=True)

# Round the 'sum' values to remove decimals and convert to integers
grouped_role['sum'] = grouped_role['sum'].round(0).astype(int)

# Plotting a bar chart
fig = px.bar(grouped_role, 
             x='sum', 
             y='TAM/IAM/ADS', 
             orientation='h', 
             title='YTD Revenue per TAM/IAM/ADS', 
             labels={'sum': 'YTD Revenue', 'TAM/IAM/ADS': 'Role'},
             width=1450, 
             height=800)

# Add annotations for YTD revenue at the end of each bar, with dots as thousand separators
for i in range(len(grouped_role)):
    fig.add_annotation(x=grouped_role['sum'][i], 
                       y=grouped_role['TAM/IAM/ADS'][i], 
                       text=f"${grouped_role['sum'][i]:,.0f}".replace(",", "."),  # Format with dots as thousand separators
                       showarrow=False, 
                       xanchor='left', 
                       yanchor='middle')

# Update the x-axis to show dots as thousand separators
fig.update_xaxes(tickformat=",.0f", tickprefix="$")  # Tick format with thousand separators and dollar sign

# Display the figure
fig.show()

# Export chart as an HTML file
fig.write_html("YTDRevenueperTAMIAMADS.html")
```

```python
# Clean and convert YTD column to numeric if not done already
ep_df['YTD'] = ep_df['YTD'].replace(',', '', regex=True).astype(float)

# Group the data by CUState and sum the YTD values, rounding to whole numbers
grouped_state = ep_df.groupby('CUState')['YTD'].sum().reset_index(name='sum').sort_values(by='sum', ascending=True)

# Round the 'sum' values to remove decimals
grouped_state['sum'] = grouped_state['sum'].round(0).astype(int)

# Plot the horizontal bar chart for YTD per state
fig = px.bar(grouped_state, 
             x='sum', 
             y='CUState', 
             orientation='h', 
             title='YTD Revenue per State', 
             labels={'sum': 'YTD Revenue', 'CUState': 'State'},
             width=1450, 
             height=1500)  

# Add annotations for YTD revenue at the end of each bar, with commas as thousand separators
for i in range(len(grouped_state)):
    fig.add_annotation(x=grouped_state['sum'][i], 
                       y=grouped_state['CUState'][i], 
                       text=f"${grouped_state['sum'][i]:,}",  # Add commas as thousand separators
                       showarrow=False, 
                       xanchor='left', 
                       yanchor='middle')

# Update the x-axis to show commas in the tick labels
fig.update_xaxes(tickformat=",.0f")  # Show commas and no decimals

# Display the figure
fig.show()

# Export chart as an HTML file
fig.write_html("YTDRevenueperState.html")
```

```python
# Replace missing values in the 'CoopCode' column with 'No CoopCode Assigned'
ep_df['CoopCode'] = ep_df['CoopCode'].fillna('No CoopCode Assigned')

# Group the data by CoopCode and sum the YTD values, including the rows with missing CoopCode
grouped_coop_code = ep_df.groupby('CoopCode')['YTD'].sum().reset_index(name='sum').sort_values(by='sum', ascending=True)

# Plotting a bar chart for YTD revenue per CoopCode
fig = px.bar(grouped_coop_code, 
             x='sum', 
             y='CoopCode', 
             orientation='h', 
             title='YTD Revenue per Coop Code (Including Blanks)', 
             labels={'sum': 'YTD Revenue', 'CoopCode': 'Coop Code'},
             width=1450, 
             height=500)

# Display the figure
fig.show()

# Export chart as an HTML file
fig.write_html("YTDRevenueperCoopCode.html")
```

```python
# Replace missing values in the 'BusinessUnitBasedOnWarehouse' and 'CUState' columns
ep_df['BusinessUnitBasedOnWarehouse'] = ep_df['BusinessUnitBasedOnWarehouse'].fillna('No Business Unit Assigned')
ep_df['CoopCode'] = ep_df['CoopCode'].fillna('No CoopCode Assigned')

# Group the data by BusinessUnitBasedOnWarehouse and CUState, and sum the YTD values
grouped_heatmap = ep_df.groupby(['BusinessUnitBasedOnWarehouse', 'CoopCode'])['YTD'].sum().reset_index()

# Scatter plot with point size representing YTD revenue
fig = px.scatter(grouped_heatmap, 
                 x='BusinessUnitBasedOnWarehouse', 
                 y='CoopCode', 
                 size='YTD', 
                 color='YTD',  # Optional: Color the points by YTD as well
                 title='Scatter Plot of YTD Revenue by Business Unit and CoopCode',
                 labels={'BusinessUnitBasedOnWarehouse': 'Business Unit', 'CoopCode': 'CoopCode', 'YTD': 'Revenue'},
                 width=1450, 
                 height=800)

# Display the scatter plot
fig.show()

# Export chart as an HTML file
fig.write_html("ScatterPlotofYTDRevenuebyBusinessUnitandCoopCode.html")
```

```python
# Group the data by BusinessUnitBasedOnWarehouse and CUState, and sum the YTD values
grouped_heatmap2 = ep_df.groupby(['BusinessUnitBasedOnWarehouse', 'CUState'])['YTD'].sum().reset_index()

# Sunburst chart for hierarchical view of YTD revenue by Business Unit and State
fig = px.sunburst(grouped_heatmap2, 
                  path=['BusinessUnitBasedOnWarehouse', 'CUState'], 
                  values='YTD', 
                  color='YTD', 
                  title='Sunburst Chart of YTD Revenue by Business Unit and State',
                  labels={'YTD': 'Revenue'},
                  width=1450, 
                  height=1200)

# Display the sunburst chart
fig.show()

# Export sunburst chart as an HTML file
fig.write_html("SunburstChartofYTDRevenuebyBusinessUnitandState.html")
```

```python
# Treemap to show YTD revenue by Business Unit and State
fig = px.treemap(grouped_heatmap2, 
                 path=['BusinessUnitBasedOnWarehouse', 'CUState'], 
                 values='YTD', 
                 color='YTD', 
                 title='Treemap of YTD Revenue by Business Unit and State',
                 labels={'YTD': 'Revenue'},
                 width=1450, 
                 height=1000)

# Display the treemap
fig.show()

# Export chart as an HTML file
fig.write_html("TreemapofYTDRevenuebyBusinessUnitandState.html")
```
